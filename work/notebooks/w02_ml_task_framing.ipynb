## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

One row = One unique content piece (content_id) aggregated over a 90-day performance window for a given client.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
else:
    # find the repo root from wherever this kernel started
    while not os.path.isdir("data/raw") and os.getcwd() != "/":
        os.chdir("..")

import pandas as pd
# Load starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create target label (1 = declining, 0 = stable/growing)
df["target"] = (df["trend_direction"] == "down").astype(int)

# Select valid features present in the dataset schema
features = [
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "avg_position",
    "ctr",
    "word_count"
]

# Output summary stats and dataframe preview
print(f"Total Rows (Pages): {len(df)}")
print(f"Unique Content Items: {df['content_id'].nunique()}")
print(f"Unique Clients: {df['client_id'].nunique()}")

# Display slice of the dataframe showing unit of analysis
df[["content_id", "client_id"] + features + ["target"]].head()
```
